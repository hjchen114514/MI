```python
def calculate_js_divergence(P, Q):
    return jensenshannon(P, Q)**2

def randomization_test_one(subject, human_sample, n1, n2, n_iterations=10000): 
    # Convert percentages to counts
    counts1 = np.round(np.array(subject) * n1 / 100).astype(int)
    counts2 = np.round(np.array(human_sample) * n2 / 100).astype(int)
    
    # actual JS divergence
    actual_divergence = calculate_js_divergence(subject, human_sample)
   
    # Combine data
    combined = np.concatenate([np.repeat(np.arange(10), counts1), np.repeat(np.arange(10), counts2)])
   
    # Randomization test
    random_divergences = []
    for _ in range(n_iterations):
        # Randomly reassign data points
        mask = bernoulli.rvs(n1 / (n1 + n2), size=len(combined))
        random_subject = np.bincount(combined[mask == 1], minlength=10)
        random_human_sample = np.bincount(combined[mask == 0], minlength=10)
       
        # Calculate JS divergence for randomized data
        random_p1 = random_subject / np.sum(random_subject)
        random_p2 = random_human_sample / np.sum(random_human_sample)
        random_divergence = calculate_js_divergence(random_p1, random_p2)
        random_divergences.append(random_divergence)
   
    # Calculate p-value
    p_value = np.mean(np.array(random_divergences) >= actual_divergence)
   
    return actual_divergence, p_value

def randomization_test_two(subject, human_sample, n1, n2, n_iterations=10000): # Two tail
    # Convert percentages to counts
    counts1 = np.round(np.array(subject) * n1 / 100).astype(int)
    counts2 = np.round(np.array(human_sample) * n2 / 100).astype(int)
    
    # Calculate actual JS divergence
    actual_divergence = calculate_js_divergence(subject, human_sample)
    
    # Combine data
    combined = np.concatenate([np.repeat(np.arange(len(subject)), counts1), 
                               np.repeat(np.arange(len(human_sample)), counts2)])
    
    # Randomization test
    random_divergences = []
    for _ in range(n_iterations):
        # Randomly reassign data points
        mask = bernoulli.rvs(n1 / (n1 + n2), size=len(combined))
        random_subject = np.bincount(combined[mask == 1], minlength=len(subject))
        random_human_sample = np.bincount(combined[mask == 0], minlength=len(human_sample))
        
        # Calculate JS divergence for randomized data
        random_p1 = random_subject / np.sum(random_subject)
        random_p2 = random_human_sample / np.sum(random_human_sample)
        random_divergence = calculate_js_divergence(random_p1, random_p2)
        random_divergences.append(random_divergence)
    
    # Calculate two-tailed p-value
    random_divergences = np.array(random_divergences)
    p_value = np.mean(np.abs(random_divergences) >= np.abs(actual_divergence))
    #p_value = np.mean(np.abs(random_divergences - np.mean(random_divergences)) >= 
    #                np.abs(actual_divergence - np.mean(random_divergences))) 
    # if each random divergence's deviation is at least as extreme as the actual divergence's deviation. This is equivalent to asking: 
    # Is this random divergence's deviation at least as extreme as the actual divergence's deviation?
    
    return actual_divergence, p_value
```

```python
def create_sample_data(percentages, n):
    """
    Create sample data based on percentages and total number of trials.
    """
    counts = np.round(np.array(percentages) * n / 100).astype(int)
    data = np.repeat(np.arange(10), counts)  # Using 11 to 20 as the range
    return data

def permutation_test_mean(group1, group2, n1, n2, n_iterations=10000):
    # Create sample data
    data1 = create_sample_data(group1, n1)
    data2 = create_sample_data(group2, n2)
    
    # Combine the data
    combined = np.concatenate([data1, data2])
    
    # Calculate the observed difference in means
    actual_diff = np.mean(data1) - np.mean(data2)
    
    # Perform permutation test
    diffs = []
    for _ in range(n_iterations):
        np.random.shuffle(combined)
        perm_data1 = combined[:len(data1)]
        perm_data2 = combined[len(data1):]
        diff = np.mean(perm_data1) - np.mean(perm_data2)
        diffs.append(diff)
    
    # Calculate two-tailed p-value
    p_value = np.mean(np.abs(diffs) >= np.abs(actual_diff))
    
    return actual_diff, p_value

def permutation_test_sd(group1, group2, n1, n2, n_iterations=10000):
    # Create sample data
    data1 = create_sample_data(group1, n1)
    data2 = create_sample_data(group2, n2)
    
    # Combine the data
    combined = np.concatenate([data1, data2])
    
    # Calculate the observed difference in standard deviations
    actual_diff = np.std(data1, ddof=1) - np.std(data2, ddof=1)
    
    # Perform permutation test
    diffs = []
    for _ in range(n_iterations):
        np.random.shuffle(combined)
        perm_data1 = combined[:len(data1)]
        perm_data2 = combined[len(data1):]
        diff = np.std(perm_data1, ddof=1) - np.std(perm_data2, ddof=1)
        diffs.append(diff)
    
    # Calculate two-tailed p-value
    p_value = np.mean(np.abs(diffs) >= np.abs(actual_diff))
    
    return actual_diff, p_value
```
